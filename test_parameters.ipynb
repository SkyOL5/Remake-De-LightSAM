```python
import torch

ckpt1 = torch.load("/home/sxcks2/Desktop/Remake-De-LightSAM/combined_model_experiments/exp_20260708_143301/model_final_0.pth", map_location="cpu")
ckpt2 = torch.load("/home/sxcks2/Desktop/Remake-De-LightSAM/combined_model_experiments/exp_20260708_143301/model_final_2.pth", map_location="cpu")
```

```python
def get_patch_decoder_params(state_dict):
    return {
        k: v
        for k, v in state_dict.items()
    }

p1 = get_patch_decoder_params(ckpt1)
p2 = get_patch_decoder_params(ckpt2)
```

```python
for k in p1.keys():
    if not torch.equal(p1[k], p2[k]):
        print(f"❌ Parameter changed: {k}")
        # break
        # continue
    else:
       print(f"✅ Parameter 完全一致，没有更新:{k}") 
# else:
#     print("✅ patch_decoder 参数完全一致，没有更新")
```
